# Dynamique 

Pour l'instant, nous n'avons fait que des résolutions "statiques", c'est-à-dire :

$$[k]\{u\} = \{f\}$$


En réalité les solides et structures vibrent et bougent en réaction aux changements
dynamiquement.

## 1) Dynamique pour un modèle masse-ressort

<center><img src="images/mass-spring.jpg"></center>

Equation du mouvement:

$$m \ddot{x}(t) = - k x(t) + F(t)$$

Equation homogène:

$$\ddot{x}(t) + \frac{k}{m} x(t) = 0$$



Solution homogène:

$$x(t) = A cos\left(\sqrt{\frac{k}{m}}t + \varphi\right)$$ 

avec:

- $\omega = \sqrt{\frac{k}{m}}$ : la pulsation naturelle
- $A$ : l'amplitude de l'oscillation
- $\varphi$ : le déphasage constant
- $A$ et $\varphi$ sont déterminés par les conditions limites et/ou initiales

La solution complète devient donc:

$$x(t) = A cos\left(\omega t + \varphi\right) + \tilde{x}(t)$$

avec $\tilde{x}(t)$ une solution particulière de l'équation non homogène :

$$m \ddot{\tilde{x} }(t) + k \tilde{x}(t) - F(t) = 0$$

## 2) Dérivation de l'équation du mouvement pour un élément barre

On reprend l'élément barre à deux noeuds (1D) sur lesquels deux forces extérieures sont appliquées $f^e_x(t)$, fonction du temps :

<center><img src="images/barre-twonodes.jpg"></center>


L'équation de Newton mais pour la mécanique des milieux continus:

$$\rho \ddot{u}(x,t) = \frac{\partial}{\partial x}\left[E \frac{\partial}{\partial x} u(x,t)\right] + T(x,t)$$

_Remarque_ :

- $\rho$ est la densité de masse $[Kg/m^3]$
- $T$ est une densité de force $[N/m^3]$
- Il s'agit de la forme forte

Avec l'interpolation des éléments finis:

$$u(x,t) = [ N(x) ] \{d(t)\}$$

_Remarque_ :

- La dépendance spatiale est portée par les fonctions d'interpolations $[N]$
- La dépendance temporelle est portée par les degrés de libertés $\{d\}$
    - $\dot{u}(x,t) = [ N(x) ] \{\dot{d}(t)\}$
    - $\ddot{u}(x,t) = [ N(x) ] \{\ddot{d}(t)\}$

Pour faire des "éléments finis" **on écrit la forme faible**:

On multiplie la forme forte par une fonction arbitraire scalaire $\color{red}{\delta u}$ et on intègre sur tout le domain (la barre et la section $A$)

$$\int_0^L A \color{red}{\delta u}^T \rho  \ddot{u}dx = \int_0^L \color{red}{\delta u}^T \frac{\partial}{\partial x}\left[EA \frac{\partial}{\partial x} u(x,t)\right]dx + \int_0^L A\color{red}{\delta u}^T T(x, t) dx $$

On intègre par partie (en supposant que $u = 0$ aux bords):

$$\int_0^L A \color{red}{\delta u}^T \rho  \ddot{u}dx = - \int_0^L \color{red}{\delta \epsilon}^T EA \frac{\partial}{\partial x} u(x,t)dx + \int_0^L A \color{red}{\delta u}^T T(x, t) dx $$

On utilise l'interpolation $\color{red}{\delta u} = [N(x)]\{\color{red}{\delta d}\}$, $\color{red}{\delta \epsilon} = [N'(x)]\{\color{red}{\delta d}\}$ et $u = [N(x)]\{d(t)\}$


$$\begin{align}
\{\color{red}{\delta d}\}^T\underbrace{\left[\int A \rho [N]^T [N]~dx\right]}_{\text{Mass Matrix}} \{\ddot{d}(t)\} =& -\{\color{red}{\delta d}\}^T \underbrace{ \left[\int [N'(x)]^T EA[N'(x)]dx \right]}_{K}\{d(t)\} \\&+ \{\color{red}{\delta d}\}^T \underbrace{ \left[\int A [N(x)]^T T(x, t) dx \right]}_{F} 
\end{align}
$$

- Comme $\{\color{red}{\delta d}\}$ est arbitraire on en déduit:

$$
\underbrace{\left[\int A \rho [N]^T [N]~dx\right]}_{\text{Mass Matrix}} \{\ddot{d}(t)\} = - \underbrace{ \left[\int [N'(x)]^T EA[N'(x)]dx \right]}_{K}\{d(t)\} + \underbrace{ \left[\int A [N(x)]^T T(x, t) dx \right]}_{F} 
$$

Soit sous forme compacte:

$$[M] \{\ddot{d}(t)\} = - [K]\{d(t)\} + \{F\} 
$$

## Matrice de masse

Tout comme la matrice de rigidité, la matrice de masse à une **equation characteristique**:

$$[M] =   \int A \rho(x) [N(x)]^T [N(x)] dx$$


Pour une barre de treillis à deux noeuds:

$$ [N(x)]^T [N(x)] = \begin{bmatrix} 1 - \frac{x}{L} \\ \frac{x}{L} \end{bmatrix} \begin{bmatrix} 1 - \frac{x}{L} & \frac{x}{L} \end{bmatrix} = 
\begin{bmatrix} \left(1 - \frac{x}{L}\right)^2 & \frac{x}{L} \left(1 - \frac{x}{L} \right) \\
\frac{x}{L} \left(1 - \frac{x}{L} \right) & \frac{x^2}{L^2}
\end{bmatrix}
$$

Les termes de la matrice de masse:

$$
\begin{align}
M_{11} &= A\rho \int \left(1-\frac{x}{L}\right)^2 dx = \rho A L \frac{1}{3}\\
M_{22} &= A\rho \int \left(\frac{x}{L}\right)^2 dx = \rho A L \frac{1}{3}\\
M_{12} &= M_{21} = A\rho \int \left(1-\frac{x}{L}\right)\frac{x}{L} dx = \rho A L \left( \frac{1}{2} - \frac{1}{3} \right) = \rho A L \frac{1}{6}
\end{align}
$$

La matrice de masse (consistante) pour une barre 1D est donc :

$$ [M] = A \rho L \begin{bmatrix} \frac{1}{3} & \frac{1}{6} \\ \frac{1}{6} & \frac{1}{3}  \end{bmatrix} =  m \begin{bmatrix} \frac{1}{3} & \frac{1}{6} \\ \frac{1}{6} & \frac{1}{3}  \end{bmatrix} $$

avec $m$ la masse totale de la barre. On remarque que $m = \sum_{ij} M_{ij}$.

*Remarque:*

- On fait (très) souvent l'approximation de condensation de la masse (lumping). 
- Cela consiste à rendre $[M]$ diagonale, en la transformat en matrice diagonale tout en conservant la masse totale du système.
- Par exemple en sommant les termes sur une ligne, 

$$M^{lumped}_{11} = M_{11} + M_{12} = \frac{A\rho L}{2}$$

et la matrice de masse (lumped) devient:

$$ [M^{lumped}] = \frac{A \rho L}{2} \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}$$

**Matrice de masse 2D**

avec une épaisseur $w$ les equations deviennent:

$$[M] =  \iint w \rho (x,y) [N(x,y)]^T [N(x,y)] dx dy$$

**Evaluation iso-paramétrique**

$$[M] =  \iint w \rho (s,t) det(J(s, t)) [N(s,t)]^T [N(s,t)] ds dt$$

**Equation du mouvement élément finis**

$$[M] \{\ddot{d}(t)\} = - [K]\{d(t)\} + \{F\} $$

Pour résoudre l'équation du mouvement, nous pouvons utiliser deux méthodes:
- Méthode de Newmark (calculer l'évolution en temps)
- Méthode des fréquences et modes propres (calcul des oscilations naturelles)

## 3) Intégration en temps: la méthode de Newmark

Pour espérer résoudre notre équation du mouvement, il nous faut discrétiser le temps. Pour ce faire on définit le déplacement au pas de temps $i$. On décompose le temps en espace de taille constante $\Delta t$, le pas de temps.

$$d_i = d(i \cdot \Delta t)$$

La méthode de Newmark-$\beta$ consiste en la définition récursive suivante sur les déplacements et les vitesses :

$$
\begin{align}
d_{i+1} &= d_i + \Delta t \dot{d}_i + \frac{\Delta t^2}{2} \left( (1 - 2\beta) \ddot{d}_i + 2\beta \ddot{d}_{i+1} \right) \\
\dot{d}_{i+1} &= \dot{d}_i + \Delta t \left( (1 - \gamma) \ddot{d}_i + \gamma \ddot{d}_{i+1} \right)
\end{align}
$$

_Remarque_:

- Le paramètre $\beta$ définit l'influence de l'accélération du passé et du futur dans le calcul du nouveau déplacement $d_{i+1}$
- dans le cas général $d_{i+1}$ dépend de $d_i$, $\dot{d}_i$, $\ddot{d}_i$ mais aussi $\ddot{d}_{i+1}$, c'est-à-dire d'une dépendance au futur.
- $\beta = 0$, $\gamma = 1/2$ : méthode des différences centrées
- $\beta = 1/4$, $\gamma = 1/2$ : règle du trapèze sans viscosité
- $\gamma > 1/2$ : introduction d'une viscosité artificielle
- $\beta = 1/2$ : une intégration de Gauss à 1 point de quadrature

Dans la suite(et pour vos projets), nous utiliserons le schéma d'intégration par différences centrées. L'algorithme prend la forme suivante:

$$
\begin{align}
1.& \quad \dot{d}_i^{\star} = \dot{d}_i + \frac{\Delta t}{2} \ddot{d}_i \\
2.& \quad d_{i+1} = d_i + \Delta t \dot{d}^{\star}_i \\
3.& \quad \ddot{d}_{i+1} = [M]^{-1}\left(-[K] d_{i+1} + F_{i+1}\right) \\
4.& \quad \dot{d}_{i+1} = d_i^{\star} + \frac{\Delta t}{2} \ddot{d}_{i+1} \\
\end{align}
$$

Remarque: Il n'y a pas de dépendance explicite au futur. On parle alors de schéma explicite

### Stabilité de la méthode de Newmark-$\beta$
 
La stabilité est une condition nécessaire et suffisante pour assurer la convergence. La méthode de Newmark-$\beta$ est dite :

- Stable inconditionellement pour: $\beta > \frac{\gamma}{2} > \frac{1}{4}$ : la règle du trapèze est toujours stable
  
- Stable sous conditions dans les autres cas :
  
  - Différences centrées: condition de Courant–Friedrichs–Lewy (CFL) :
  
   $$\Delta t < \alpha \frac{h}{c}$$ 

  avec 
  
  - 1D: $c = \sqrt{\frac{E}{\rho}}$, $h$ la taille de maillage et $\alpha$ un facteur de sécurité ($\sim 10\%$)
  - 3D: $c = c_p = \sqrt{\frac{\lambda+2\mu}{\rho}}$ (pressure wave), $h$ la taille du cercle inscrit dans un élément de maillage et $\alpha$ un facteur de sécurité ($\sim 10\%$)

### Mesure de l'énergie

La mesure de l'énergie est un des moyens indispensables pour vérifier la bonne **stabilité** du code. Lorsqu'il n'y a aucun travail extérieur appliqué l'énergie totale doit être conservée:

$$E^{tot} = E^{cin} + E^{pot}$$

Avec:
- l'énergie potentielle: $$E^{pot} = \frac{1}{2} \{d\}^T [K] \{d\}$$
- l'énergie cinétique : $$E^{cin} = \frac{1}{2} \{\dot{d}\}^T [M] \{\dot{d}\}$$

## 4) Fréquences naturelles et modes propres

Nous avons vu que l'équation du mouvement prenait la forme:

$$[M]\{\ddot{d}\} = - [K] \{d\} + \{f\}$$

Pour calculer les pulsations naturelles, on va supposer que les déplacements s'écrivent :

$$\{d(t)\} = e^{i\omega t} \{d'\}$$

où :
- $\{d'\}$ est un vecteur nodal qui ne dépend pas du temps
- $e^{i\omega t}$ est un scalaire complexe

Dès lors, on peut écrire:

$$\{\ddot{d}(t)\} = - \omega^2 \{d(t)\} = - \omega^2 e^{i\omega t} \{d'\}$$

Et l'équation du mouvement homogène 

$$[M]\{\ddot{d}\} + [K] \{d\} = 0$$

devient:

$$- \omega^2 [M]\{d'\} + [K] \{d'\} = 0$$

Soit encore :

$$\left(- \omega^2 [M] + [K]\right)\{d'\} = 0$$



Cela revient à résoudre un problème aux valeurs et vecteurs propres. En effet, si ($\lambda,\{v\}$) est une paire de valeur et de vecteur propres de la matrice $[M]^{-1}[K]$
alors:

$$[M]^{-1}[K] \{v\} = \lambda \{v\}$$

soit:

$$[K] \{v\} = \lambda [M]\{v\}$$

et donc $\lambda = \omega^2$

Il y a deux ingrédients: les vecteurs propres $\{v\}$, les fréquences $2\pi f = \omega$ et la période $T = \frac{1}{f} = \frac{2\pi}{\omega}$

Comme les vecteur propres ($\{v_j\}$) forment une base orthogonale (que l'on peut normaliser) de l'espace, n'importe quel vecteur de déplacement $\{d(t)\}$ peut s'écrire:

$$d(t) = \sum_{j=1}^N A_j v_j e^{i\omega_j t}$$

avec $A_j$ un nombre complexe. N est le nombre de nœuds correspondant au nombre de modes. Cela peut s'écrire 
matriciellement par:

$$\{d(t)\} = [V] \{A(t)\}$$ 

avec $[V]$ la matrice des vecteurs propres normalisée (en colonne) 
et $$
\{A(t)\}_j = \left\{
    \begin{array}{ll}
        A_j e^{i\omega_j t} & \text{si } t > 0  \\
        A_j & \text{si } t = 0
    \end{array}
\right.
$$

Parce que $[V]$ est une matrice orthogonale, on peut alors trouver complétement les déplacements grâce aux conditions initiales :

$$[V]^T [M]\{d(0)\} = \{A(0)\}$$ 

En effet le produit scalaire selon lequel les vecteurs propres sont orthogonaux est:

$<v_i, v_j> = v_i^T M v_j = \delta_{ij}$

donc:

$$[V]^T [M] [V] \{A(t)\} = [I_d] \{A(0)\} = \{A(0)\}$$

### Importance des modes propres et fréquences propres

Puisque l'on peut décomposer tout mouvement sur la base de vecteurs propres, on peut caractériser toutes les évolutions dynamiques à l'aide des fréquences et des vecteurs propres associés.

On peut caractériser les contraintes maximales pour chacun de ces modes. La plupart des études de la dynamique, telles que les analyses de risque des tremblements de terre, tirent partis de cette approche. En revanche, le matériau doit être élastique.
