# Conditionnement d'une matrice

Le conditionnement d'un problème est une mesure de la sensitivité du résultat sur de petites variations des données. 

Pour le problème de la résolution du système d'équations linéaires $Ax=b$, on peut définir une pertubation de $b$ en $\hat{b}$ et le vecteur de pertubation $\Delta b=\hat{b} -b$. Cette perturbation de $b$ entraîne une variation $\Delta x$ de $x$ :

\begin{equation}
A(x+\Delta x) = b + \Delta b,
\end{equation}

\begin{equation}
Ax + A\Delta x  = Ax +\Delta b ,
\end{equation}

\begin{equation}
A \Delta x  = \Delta b.
\end{equation}

Le conditionnement de la matrice $A$ est défini comme le maximum du ratio entre l'erreur relative en $x$ : $||\Delta x||/||x||$  et l'erreur relative en $b$ : $||\Delta b||/||b||$. Le ratio peut être exprimé comme suit:


\begin{equation}
\Big[ \frac{||\Delta x||}{||x||} \frac{||b||}{||\Delta b||} \Big] = \Big[ \frac{||A^{-1}\Delta b||}{||A^{-1}b|||} \frac{||b||}{||\Delta b||} \Big],
\end{equation}

avec $|| \cdot ||$  une norme.


Et donc le contionnement $\kappa(A)$ de $A$ comme :


\begin{equation}
\kappa(A) = max_{b,\Delta b \neq 0} \Big[ \frac{||A^{-1}\Delta b||}{||A^{-1}b|||} \frac{||b||}{||\Delta b||} \Big] =
max_{\Delta b \neq 0} \Big[ \frac{||A^{-1}\Delta b||}{||\Delta b||} \Big] max_{b \neq 0} \Big[ \frac{||b||}{||A^{-1}b|||} \Big] =
max_{\Delta b \neq 0} \Big[ \frac{||A^{-1}\Delta b||}{||\Delta b||} \Big] max_{x \neq 0} \Big[ \frac{||Ax||}{||x||} \Big].
\end{equation}

\begin{equation}
\kappa(A) =
max_{\Delta b \neq 0} \Big[ \frac{||A^{-1}\Delta b||}{||\Delta b||} \Big] max_{x \neq 0} \Big[ \frac{||Ax||}{||x||} \Big].
\end{equation}


Si $|| \cdot ||$ est une norme consistente, alors on peut écrire:

\begin{equation}
\kappa(A) = ||A^{-1}||\, ||A||.
\end{equation}


Et enfin, si on considère une norme euclidienne on a :

\begin{equation}
\kappa(A) = \frac{|\lambda_{max}|}{|\lambda_{min}|},
\end{equation}


avec $\lambda_{max}$ et $\lambda_{min}$ la plus grande et la plus petite valeur propre de $\mathbf{A}$ (en valeur absolue), et $| \cdot |$ l'opérateur de valeur absolue.

Le conditionnement $\kappa(A)$ donne l'amplification maximum d'une erreur en $b$ dans le résultat $x$:

\begin{equation}
\frac{||\Delta x||}{||x||} \leq \kappa(A) \frac{||\Delta x||}{||x||}
\end{equation}


On peut aussi montrer une relation similaire pour le cas d'une pertubation $\Delta A$ de la matrice $A$:

\begin{equation}
\frac{||\Delta x||}{||x+ \Delta x||} \leq \kappa(A) \frac{||\Delta A||}{||A||}
\end{equation}

Considérons une structure composée de deux barres horizontales, encastrée à gauche et soumis à une force F=1 à droite. Les deux barres ont respectivement une rigidité de $k_1 = \frac{EA_1}{L_1}$ et $k_2 = \frac{EA_2}{L_2}$ et la longueur totale est fixée à 1.

La matrice de rigidité de la structure $K_{tot}$ est donc :

\begin{equation}
K_{tot} = \begin{bmatrix}
k_1  & -k_1 & 0 \\
-k_1 & k_1 + k_2 & -k_2 \\
0 & -k_2 & k_2
\end{bmatrix}.
\end{equation}

En appliquant les conditions de bord, on doit résoudre le système suivant :

\begin{equation}
\mathbf{Kd=F}
\end{equation}

\begin{equation}
\begin{bmatrix}
k_1 + k_2 & -k_2 \\
-k_2 & k_2
\end{bmatrix}
\begin{bmatrix}
d_1 \\
d_2
\end{bmatrix}
=
\begin{bmatrix}
0 \\
F
\end{bmatrix}
\end{equation}

Pour étudier l'effet du condionnement de $\mathbf{K}$, le système est aussi résolu avec une matrice de rigidité avec une petite perturbation $\mathbf{K^{pert}}$:

\begin{equation}
\begin{bmatrix}
1.000001(k_1 + k_2) & -k_2 \\
-k_2 & k_2
\end{bmatrix}
\begin{bmatrix}
d^*_1 \\
d^*_2
\end{bmatrix}
=
\begin{bmatrix}
0 \\
F
\end{bmatrix}
\end{equation}


En dessous, il est possible de modifier la longueur de $L1$ et la rigidité $EA$ des deux barres et d'obtenir directement le conditionnement du systême ainsi que la solution du système normal $\mathbf{d}$ et perturbé $\mathbf{d^*}$.
